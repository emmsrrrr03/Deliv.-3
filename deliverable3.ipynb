```python
import random

class Character:

    def __init__(
        self,
        name: str,
        HP_max: int = 100,
        damage: int = 10,
        weapon=None
    ):

        self.name = name
        self._HP = HP_max
        self._max_HP = HP_max
        self.damage = damage
        self.weapon = weapon

    @property
    def is_alive(self):
        return self._HP > 0

    @property
    def HP(self):
        return self._HP

    def take_damage(self, amount):

        self._HP -= amount

        if self._HP < 0:
            self._HP = 0

    def heal(self, amount):

        self._HP += amount

        if self._HP > self._max_HP:
            self._HP = self._max_HP

    def attack(self, target):

        if self.weapon:
            damage = self.weapon.roll_damage()
        else:
            damage = self.damage

        target.take_damage(damage)
      
```

```python
class Dice:

    def __init__(self, sides=6):
        self.sides = sides

    def roll(self):
        return random.randint(1, self.sides)

    def __call__(self):
        return self.roll()

    def __repr__(self):
        return f"Dice(sides={self.sides})"
```

```python
class Warrior(Character):

    def __init__(
        self,
        name,
        hp,
        armor=5,
        weapon=None
    ):

        super().__init__(
            name=name,
            HP_max=hp,
            weapon=weapon
        )

        self.armor = armor

    def take_damage(self, amount):

        reduced_damage = max(
            0,
            amount - self.armor
        )

        super().take_damage(reduced_damage)
```

```python
class Mage(Character):

    def __init__(
        self,
        name,
        hp,
        mana=30,
        mana_max=30,
        weapon=None
    ):

        super().__init__(
            name=name,
            HP_max=hp,
            weapon=weapon
        )

        self.mana = mana
        self.mana_max = mana_max

    def attack(self, target):

        if self.mana >= 10:

            self.mana -= 10

            damage = self.damage * 2

            target.take_damage(damage)

        else:

            self.mana = min(
                self.mana_max,
                self.mana + 5
            )

            if self.weapon:
                damage = self.weapon.roll_damage()
            else:
                damage = self.damage

            target.take_damage(damage)
```

```python
class Battle:

    def __init__(self, hero, enemy):

        self.hero = hero
        self.enemy = enemy

    def play_round(self):

        if self.hero.is_alive and self.enemy.is_alive:
            self.hero.attack(self.enemy)

        if self.enemy.is_alive:
            self.enemy.attack(self.hero)

    def simulate_battle(self):

        round_number = 1

        while self.hero.is_alive and self.enemy.is_alive:

            print(f"\n--- Round {round_number} ---")

            self.play_round()

            print(
                f"{self.hero.name} HP: {self.hero.HP}"
            )

            print(
                f"{self.enemy.name} HP: {self.enemy.HP}"
            )

            round_number += 1

        print()

        if self.hero.is_alive:
            print("Hero has won the match!")
        else:
            print("Enemy has won the match!")
```
