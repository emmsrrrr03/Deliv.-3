```python
#TASK 1
import random

class Character:
#Initialize the caracter
    def __init__(
        self,
        name: str,
        HP_max: int = 100,
        damage: int = 10,
        weapon=None
    ):

        self.name = name
        self._HP = HP_max
        self._max_HP = HP_max
        self.damage = damage
        self.weapon = weapon
#check if the character is alve
    @property
    def is_alive(self):
        return self._HP > 0
#return current hp
    @property
    def HP(self):
        return self._HP
#reduce hp when taking damage
    def take_damage(self, amount):

        self._HP -= amount

        if self._HP < 0:
            self._HP = 0
#increase damage when healed
    def heal(self, amount):

        self._HP += amount

        if self._HP > self._max_HP:
            self._HP = self._max_HP
#attacjk target
    def attack(self, target):

        if self.weapon:
            damage = self.weapon.roll_damage()
        else:
            damage = self.damage

        target.take_damage(damage)
      
```

```python
#TASK 2
#importing Dice with 6 sides
class Dice:

    def __init__(self, sides=6):
        self.sides = sides

    def roll(self):
        return random.randint(1, self.sides)

    def __call__(self):
        return self.roll()

    def __repr__(self):
        return f"Dice(sides={self.sides})"
```

```python
#TASK 3 defining weapon 
class Weapon:
    def __init__(self, name, damage_dice):
        self.name = name
        self.damage_dice = damage_dice
    def roll_damage(self):
        return self.damage_dice.roll()
```

```python
#TASK 4 added characters and defined hp levels and attack strenght
class Warrior(Character):

    def __init__(
        self,
        name,
        hp,
        armor=5,
        weapon=None
    ):

        super().__init__(
            name=name,
            HP_max=hp,
            weapon=weapon
        )

        self.armor = armor

    def take_damage(self, amount):

        reduced_damage = max(
            0,
            amount - self.armor
        )

        super().take_damage(reduced_damage)
```

```python
#Task 4
class Mage(Character):

    def __init__(
        self,
        name,
        hp,
        mana=30,
        mana_max=30,
        weapon=None
    ):

        super().__init__(
            name=name,
            HP_max=hp,
            weapon=weapon
        )

        self.mana = mana
        self.mana_max = mana_max

    def attack(self, target):

        if self.mana >= 10:

            self.mana -= 10

            damage = self.damage * 2

            target.take_damage(damage)

        else:

            self.mana = min(
                self.mana_max,
                self.mana + 5
            )

            if self.weapon:
                damage = self.weapon.roll_damage()
            else:
                damage = self.damage

            target.take_damage(damage)
```

```python
#TASK 5
class Battle:
#defining batlle and how the battle is played out. 
    def __init__(self, hero, enemy):

        self.hero = hero
        self.enemy = enemy

    def play_round(self):

        if self.hero.is_alive and self.enemy.is_alive:
            self.hero.attack(self.enemy)

        if self.enemy.is_alive:
            self.enemy.attack(self.hero)

    def simulate_battle(self):

        round_number = 1

        while self.hero.is_alive and self.enemy.is_alive:

            print(f"\n--- Round {round_number} ---")

            self.play_round()

            print(
                f"{self.hero.name} HP: {self.hero.HP}"
            )

            print(
                f"{self.enemy.name} HP: {self.enemy.HP}"
            )

            round_number += 1

        print()

        if self.hero.is_alive:
            print("Hero has won the match!")
        else:
            print("Enemy has won the match!")
```

```python
d6 = Dice(6)

sword = Weapon(
    "Sword",
    d6
)

me = Warrior(
    name="Me",
    hp=100,
    weapon=sword
)

you = Mage(
    name="You",
    hp=80
)

battle = Battle(
    hero=me,
    enemy=you
)
```

```python
battle.simulate_battle()
```

In this project we built a turn-based RPG battle system using Object-Oriented Programming. Over the course of five tasks, we designed characters, dice rolling mechanics, weapons, unique subclasses like warriors and mages, and a complete battle simulation loop. 


Task 1: 

the Character class serves as the blueprint for all fighters in our game. In the constructor, we initialize attributes like name, HP levels, damage, and an optional weapon. We also use properties like is_alive to check if a character's health is above zero, and methods like take_damage and heal to manage their health pool safely, making sure health never drops below zero or exceeds the maximum limit. Finally, the attack method handles striking a target, factoring in whether the character has a weapon equipped.



Task 2 and 3: we added randomness to our combat. Here we built a Dice class using Python's random module, allowing us to roll dice with any number of sides. 

Then, in Task 3, we created a Weapons class. A weapon takes a name and a damage dice object. When a character attacks with a weapon, it calls roll_damage() to generate dynamic, randomized damage instead of a fixed number.


Moving on to Task 4, we used inheritance to create specialized character types: the Warrior and the Mage. The Warrior adds an armor attribute. When the warrior takes damage, the armor reduces the incoming hit, making them much sturdier in battle. The Mage, on the other hand, manages a mana pool. When the mage attacks, if they have enough mana, they consume it to cast a powerful magical spell that deals double base damage. If they run low on mana, they automatically regenerate some mana and fall back to a standard attack


In Task 5, we brought everything together with the Battle class. This class takes a hero and an enemy and manages the combat flow. The play_round method checks if both fighters are alive, lets the hero attack, and then lets the enemy counter-attack if they survived. The simulate_battle method wraps this in a while loop that runs until one character is defeated, and finally prints out who won the match.
