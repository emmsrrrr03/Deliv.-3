```python
class Warrior(Character):
    def __init__(self, name: str, hp: int, armor: int = 5, weapon=None):
        super().__init__(name, hp, weapon)
        self.armor = armor

    def take_damage(self, amount: int) -> None:
        #reduser skade med armor, men ikke for mye
        reduced_damage = max(0, amount - self.armor)
        super().take_damage(reduced_damage)


class Mage(Character):
    def __init__(self, name: str, hp: int, mana: int = 30, mana_max: int = 30, weapon=None):
        super().__init__(name, hp, weapon)
        self.mana = mana
        self.mana_max = mana_max

    def attack(self, target: Character) -> None:
        if self.mana >= 10:
            self.mana -= 10
            #magisk angrep med dobbel skade (uten våpen)
            base_damage = 5  # eller standardbase fra Character
            damage = base_damage * 2
            target.take_damage(damage)
        else:
            #gjenopprett 5 mana ved fallback, og vanlig angrep med våpen
            self.mana = min(self.mana_max, self.mana + 5)
            super().attack(target)
```
