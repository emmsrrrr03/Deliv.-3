```python
# Task 1
class Character:

    #Initialize the caracter
    def __init__(self, name: str, HP_max = 100, damage = 10, weapon = None) -> None:
        self.name = name
        self._HP = HP_max
        self._max_HP = HP_max
        self.damage = damage
        self.weapon = weapon

#check if the character is alive
    @property
    def is_alive(self) -> bool:
        return self._HP > 0

#return current HP
    @property
    def HP(self) -> int:
        return self._HP
    
#reduce HP when taking damage
    def take_damage(self, amount) -> None:
        self._HP -= amount
        if self._HP <= 0:
            self._HP = 0

#Increase HP when heald     
    def heal(self, amount) -> None:
        self._HP += amount
        if self._HP > self._max_HP:
            self._HP = self._max_HP

#attac target
    def attack(self, target) -> None:
        if self.weapon:
            damage = self.weapon.roll_damage()
        else:
            damage = self.damage
            
        target.take_damage(damage)
```

```python
# Task 2
import random

class Dice:
    def __init__(self, sides: int = 6):
        self.sides = sides

    def roll(self) -> int:
        return random.randint(1, self.sides)

    def __call__(self) -> int:
        return self.roll()

    def __repr__(self) -> str:
        return f"Dice (sides={self.sides})"
```
